```python
import asyncio
import time


async def task_a():
    print("Task A started")
    await asyncio.sleep(2)
    print("Task A finished")

    return "Result A"


async def task_b():
    print("Task B started")
    await asyncio.sleep(1)
    print("Task B finished")

    return "Result B"


async def main():
    results = await asyncio.gather(
        task_a(),
        task_b(),
    )

    print(results)


start = time.perf_counter()

await main()

# asyncio.run(main())

print(f"Time taken: {time.perf_counter() - start:.2f} seconds")
```

```python
async def generate_numbers():
    for number in range(5):
        await asyncio.sleep(2)
        yield number


async def main():

    async for number in generate_numbers():
        print(number)


await main()
```

```python
import asyncio
import httpx


async def fetch_post():

    async with httpx.AsyncClient() as client:

        response = await client.get(
            "https://jsonplaceholder.typicode.com/posts/1"
        )

        return response.json()


async def main():
    post = await fetch_post()
    print(post)


await main()
```

```python
import asyncio
import httpx


async def fetch_user(client):
    response = await client.get(
        "https://jsonplaceholder.typicode.com/users/1"
    )

    return response.json()


async def fetch_post(client):
    response = await client.get(
        "https://jsonplaceholder.typicode.com/posts/1"
    )

    return response.json()


async def fetch_todo(client):
    response = await client.get(
        "https://jsonplaceholder.typicode.com/todos/1"
    )

    return response.json()


async def main():
    
    async with httpx.AsyncClient() as client:

        user, post, todo = await asyncio.gather(
            fetch_user(client),
            fetch_post(client),
            fetch_todo(client),
        )

    print("User:")
    print(user["name"])

    print("\nPost:")
    print(post["title"])

    print("\nTodo:")
    print(todo["title"])


await main()
```
