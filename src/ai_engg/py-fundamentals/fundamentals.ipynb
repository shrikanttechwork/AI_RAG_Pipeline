## 2. Basic Types and `None`

Python includes built-in types such as `str`, `int`, `float`, `bool`, and `NoneType`[cite: 1].

### The `None` Gotcha for Engineers
`None` is Python's singleton representing the absence of a value (similar to `null` or `nil`)[cite: 1]. 

| Check Syntax | Meaning | Evaluates to True For |
| :--- | :--- | :--- |
| `value is None` | Identity check (Is this object literally `None`?)[cite: 1] | `None`[cite: 1] |
| `not value` | Truthiness check (Is this value falsy?)[cite: 1] | `None`, `0`, `""`, `[]`, `{}`, `False`[cite: 1] |

> **Crucial Rule:** Always check for missing values using `value is None` rather than `not value` to avoid treating `0` or empty strings as missing[cite: 1].

````python
name = "Anuj"            # str
age = 28                 # int
price = 99.5             # float
is_active = True         # bool
value = None             # represents "no value"

print(f"types -> name:{type(name)}, age:{type(age).__name__}, "
      f"price:{type(price).__name__}, is_active:{type(is_active).__name__}, "
      f"value:{type(value).__name__}")

value = 0
print(f"value is None -> {value is None}   (False — 0 is a real value)")
print(f"not value      -> {not value}       (True  — 0 is falsy)")
````

## 4. Control Flow

Python relies on **indentation** (4 spaces) rather than curly braces `{}` to define code blocks[cite: 1].

### Loop Constructs
1. `for item in iterable`: Direct iteration over items (preferred over index tracking)[cite: 1].
2. `for i in range(n)`: Generates numbers from `0` to `n-1`[cite: 1].
3. `while condition`: Standard conditional loop[cite: 1].
4. `continue` & `break`: Control loop execution flow[cite: 1].

````python
age = 8
if age >= 18:
    status = "adult"
elif age >= 13:
    status = "teen"
else:
    status = "child"
print(f"status -> {status}")
````

````python
users = ["Anuj", "Rahul", "Amit"]

print("for-each over a list:")
for user in users:
    print(f"   - {user}")

print("classic counting loop with range():")
for i in range(3, 5):
    print(f"   i = {i}")

print("while loop:")
retries = 0
while retries < 3:
    retries += 1
    print(f"   retry attempt {retries}")

print("continue / break:")
items = ["a", None, "b", "stop", "c"]
for item in items:
    if item is None:
        continue
    if item == "stop":
        break
    print(f"   processing {item}")
````

## 5. Core Collections

Understanding Python's standard data structures is vital for handling payloads, data structures, and feature vectors[cite: 1].

| Collection | Syntax | Ordered? | Mutable? | Unique Elements Only? |
| :--- | :--- | :--- | :--- | :--- |
| `list` | `[1, 2, 3]` | Yes | Yes | No[cite: 1] |
| `tuple` | `(1, 2, 3)` | Yes | No | No[cite: 1] |
| `dict` | `{"key": "val"}` | Yes (Insertion order) | Yes | Unique Keys[cite: 1] |
| `set` | `{1, 2, 3}` | No | Yes | Yes[cite: 1] |

````python
# list — ordered, mutable
users = ["Anuj", "Rahul"]
users.append("Amit")
print(f"list -> {users}, first item -> {users[0]}")

# dict — key/value pairs
user = {"name": "Anuj", "age": 28}
print(f"dict -> {user['name']}")
print(f".get on missing key       -> {user.get('email')}")          # Returns None, avoids KeyError
print(f".get with default         -> {user.get('email', 'unknown')}")

for key, value in user.items():
    print(f"   {key} = {value}")

# tuple — ordered, immutable
point = (10, 20)
print(f"tuple -> {point}")

# set — unordered, unique values only
tags = {"agent", "rag", "agent"}     # Duplicates are automatically removed
print(f"set (duplicates collapsed) -> {tags}")
````

## 6. Functions & Type Hints

Functions are declared with `def`. Modern Python includes **type hints** (`name: str -> str`), which help IDEs and static analysis tools[cite: 1].

> **Important:** Python type hints are **not enforced at runtime** by the interpreter[cite: 1]. Libraries like *Pydantic* or *MyPy* are used to perform runtime checking or static verification[cite: 1].

````python
def greet(name: str, greeting: str = "Hello") -> str:
    return f"{greeting}, {name}"

print(greet('Anuj'))                 # Default greeting
print(greet('Anuj', 'Hi'))           # Override default
print(greet(name='Anuj'))            # Keyword argument

# Type hints are NOT enforced at runtime:
print(f"type hints are NOT enforced -> {greet(123)}")
````

````python
# Reference demonstration
a = [1, 2, 3]
b = a              # b references the same object as a
b.append(4)
print(f"a -> {a}   (a changed because b points to the same object)")
print(f"a is b -> {a is b}")

# Immutable types do not show this behavior
x = 4 #memory at 100
y = x #memory at 200 
x = 5              # Reassigns x to a new integer object
print(f"x={x}, y={y}   (y remains 4)")
````

````python
# Creating custom exceptions for API errors / validation
class APIValidationError(Exception):
    """Custom exception raised when API validation fails."""
    pass

def process_api_payload(payload: dict):
    if "prompt" not in payload:
        raise APIValidationError("Missing required 'prompt' key in payload.")
    return f"Processing prompt: {payload['prompt']}"

# Full Exception Handling Lifecycle
def execute_pipeline(payload: dict):
    try:
        print("Starting pipeline execution...")
        result = process_api_payload(payload)
    except APIValidationError as err:
        print(f"Caught specific API Validation Error: {err}")
    except Exception as err:
        print(f"Caught unexpected generic error: {err}")
    else:
        print(f"Pipeline succeeded: {result}")
    finally:
        print("Cleanup: closing connections and flushing logs...")

# print("--- Executing with invalid payload ---")
# execute_pipeline({})

print("\n--- Executing with valid payload ---")
execute_pipeline({"prompt": "Summarize this text"})
````

## 10. List Comprehensions

List comprehensions provide a clean syntax for creating lists based on existing iterables[cite: 1].

### General Syntax Patterns
1. **Transformation:** `[expression for item in iterable]`[cite: 1]
2. **Filtering:** `[expression for item in iterable if condition]`[cite: 1]

> **Resource:** Learn more about [Python List Comprehensions](https://www.w3schools.com/python/python_lists_comprehension.asp).

````python
numbers = [1, 2, 3, 4, 5]

# Transformation
squares = [x * x for x in numbers]
print(f"squares -> {squares}")

# Filtering
even = [x for x in numbers if x % 2 == 0]
print(f"even    -> {even}")

def hi(i):
    print("hi", i)

a = [hi(i) for i in range(1, 3)]
print(a)
````
