#### Problem 1.1: Type Casting & Truthiness Evaluation

Create a function or script that receives three raw inputs:

* `raw_price`: string (e.g., `"199.99"`)
* `raw_quantity`: string (e.g., `"5"`)
* `raw_discount`: string (e.g., `"0"`)

**Requirements:**

1. Cast `raw_price` to a `float` and `raw_quantity` to an `int`.
2. Compute `total_cost = price * quantity`.
3. Evaluate the truthiness of `raw_discount`. If `raw_discount` is non-zero, calculate a discounted price; otherwise, keep `total_cost` as is.
4. Verify whether `total_cost is None` vs `not total_cost`. Write comments explaining the difference between these two checks.

````python
def calculate_discounted_price(raw_price: str, raw_quantity: str, raw_discount: str):
    price = float(raw_price)
    quantity = int(raw_quantity)

    total_cost = price * quantity

    discount = float(raw_discount)

    if discount:
         final_price = total_cost * (1 - discount / 100)
    else:
         final_price = total_cost

    print("Total Cost:", total_cost)
    print("Final Price:", final_price)

    if total_cost is None:
         print("total_cost is None")

    if not total_cost:
         print("total_cost is falsy")
````

````python
calculate_discounted_price("100.50", "5", "10")
````

### Problem 2.1: Dynamic Prompt Generator

Build a string-processing function `generate_llm_prompt()` that formats incoming system instructions and dynamic user inputs for an AI API.

**Requirements:**

1. Accept three parameters: `system_role` (str), `user_query` (str), and `temperature` (float).
2. Clean `user_query` by removing leading/trailing whitespaces.
3. Use an **f-string** with specific formatting specs:
* Format `temperature` to exactly 2 decimal places (e.g., `0.70`).
* Convert `system_role` to uppercase.
4. Capitalize the first letter of `user_query`.

**Expected Input:**

```python
generate_llm_prompt("  system assistant  ", "   what is rag?   ", 0.7)

```

**Expected Output:**

```text

```

````python
def generate_llm_prompt(system_role: str, user_query: str, temperature: float):
    return f"[SYSTEM: {system_role.strip().replace("system", "").strip().upper()} | TEMP: {temperature:.2f}] {user_query.strip().capitalize()}"

result = generate_llm_prompt("  system assistant  ", "   what is rag?   ", 0.7)
print(result)
````

### Problem 3.1: API Retry Counter with Exponential Backoff

Simulate an API request retry handler using a loop.

**Requirements:**

1. Given a list of simulated server responses: `responses = [500, 502, 503, 200, 404]`.
2. Iterate through `responses` using a `while` loop or `for` loop:
* If the code is `500`, `502`, or `503`, print a warning message indicating a retry, increment the attempt counter, and double the wait time (start backoff at 1 second).
* If the code is `200`, print success and `break` out of the loop immediately.
* For any other error (e.g., `404`), print a fatal error and `break`.


3. Skip `None` or invalid values in the list using `continue`.

````python
responses = [500, 502, 503, 200, 404]
attempt = 1
backoff = 1

for code in responses:
    if code is None or not isinstance(code, int):
        continue
    if code in (500, 502, 503):
        print(
            f"Attempt {attempt}: Server error {code}. "
            f"Retrying in {backoff} second(s).."
            )
        attempt += 1
        backoff *= 2
    elif code == 200:
        print(f"Attempt {attempt}: Success! Server responded with {code}.")
        break
    else:
        print(f"Attempt {attempt}: Fatal error {code}. Stopping.")
        break
````

chat_logs = [
    {"user": "Alice", "tags": ["python", "ai"], "msg": "Hello world"},
    {"user": "Bob", "tags": ["rag", "ai"], "msg": "Building a RAG pipeline"},
    {"user": "Alice", "tags": ["python"], "msg": "Learning collections"},
]

```

**Requirements:**

1. Create a `set` containing all **unique tags** across all chat messages.
2. Build a `dict` mapping each `user` to a `list` of their messages.
3. Access missing keys safely using `.get()` without raising a `KeyError`.

**Expected Outputs:**

```python
# Unique Tags Set:
{"python", "ai", "rag"}

# User Index Dict:
{
    "Alice": ["Hello world", "Learning collections"],
    "Bob": ["Building a RAG pipeline"]
}


````python
chat_logs = [
    {"user": "Alice", "tags": ["python", "ai"], "msg": "Hello world"},
    {"user": "Bob", "tags": ["rag", "ai"], "msg": "Building a RAG pipeline"},
    {"user": "Alice", "tags": ["python"], "msg": "Learning collections"},
]

unique_tags = set()
user_messages = {}

for log in chat_logs:
    # Req 1
    unique_tags.update(log.get("tags", []))

    # Req 2
    user = log.get("user")
    msg = log.get("msg")
    user_messages[user] = user_messages.get(user, [])
    user_messages[user].append(msg)

# Req 3
charlie_messages = user_messages.get("Charlie", [])

print("Unique Tags:", unique_tags)
print("User Messages:", user_messages)
print("Charlie's Messages:", charlie_messages)
````

````python
from datetime import datetime
from typing import Optional

def add_log_entry(event: str, level: str = "INFO", logs: Optional[list[str]] = None) -> list[str]:
    """
    Safely adds a log entry to the log list without mutable default bugs.
    """
    if logs is None:
        logs = []

    timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    log_entry = f"[{timestamp}] [{level}] {event}"

    logs.append(log_entry)

    return logs
````
