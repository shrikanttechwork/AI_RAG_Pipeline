```python
class Dog:

    species = "Dog"

    def __init__(self, name: str, breed: str):
        self.name = name
        self.breed = breed

    def bark(self):
        self.eyes = 2
        return f"{self.name} says woof and has {self.eyes} eyes!"

    @classmethod
    def walk(cls):
        print(f"{cls.species} is walking")
```

```python
leo = Dog(name="Leo", breed="Labra")
print(leo.bark())
print(leo.eyes)
print(leo.breed)
leo.walk()
```

```python
class User:

    def __init__(self, name):
        self.name = name

    @classmethod
    def from_dict(cls, data):
        return cls(data["name"])


user = User.from_dict({"name": "shrikant"})
print(user.name)
```

```python
class MathUtils:

    @staticmethod
    def add(a, b):
        return a + b

mathutil = MathUtils.add(2, 3)

print(mathutil)
```

```python
class UserInfo:

    def __init__(self, first_name, last_name):
        self.first_name = first_name
        self.last_name = last_name

    @property
    def full_name(self):
        return f"{self.first_name} {self.last_name}"
```

```python
user = UserInfo("Test1", "Test2")
print(user.full_name)
```
